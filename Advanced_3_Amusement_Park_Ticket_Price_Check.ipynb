```python
def calculate_ticket_price(age):
    """Returns the ticket price based on age."""
    if age < 12:
        return 5
    elif age <= 18:
        return 15
    else:
        return 20


def add_picture_charge(bill, picture_choice):
    """Adds picture charge if required."""
    if picture_choice.lower() == 'y':
        bill += 3
    return bill


def main():
    print("Welcome to the Amusement Park!\n")

    height = float(input("Please enter your height in cm: "))

    if height > 120:
        print("You are eligible for games and rides!")

        age = int(input("Please enter your age: "))

        bill = calculate_ticket_price(age)
        print(f"Your ticket price is ${bill}")

        want_picture = input(
            "Do you need a picture? Type y for Yes and n for No: "
        )

        bill = add_picture_charge(bill, want_picture)

        print(f"Your final bill is ${bill}")

    else:
        print("You are not eligible for games and rides!")


# Program execution starts here
main()
```
