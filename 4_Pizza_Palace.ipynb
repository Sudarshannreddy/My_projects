```python
print('Welcome to Pizza Palace!')

bill = 0

size = input('Enter the size of pizza you need S, M or L: ').upper()
macaroni = input('Do you need macaroni? Type Y for Yes and N for No: ').upper()
extra_cheese = input('Do you need extra cheese? Type Y for Yes and N for No: ').upper()

if size == 'S':
    bill = 20
elif size == 'M':
    bill = 25
elif size == 'L':
    bill = 30
else:
    print("You have entered an incorrect size.")

if macaroni == 'Y':
    bill += 3

if extra_cheese == 'Y':
    bill += 5

print(f'Your final bill is ${bill}')
```
