```python
df['day']=df['Date_of_Journey'].str.split('/').str[0]
df['month']=df['Date_of_Journey'].str.split('/').str[1]
df['year']=df['Date_of_Journey'].str.split('/').str[2]
df.drop('Date_of_Journey',axis=1,inplace=True)
df['day']=df['day'].astype(int)
df['month']=df['month'].astype(int)
df['year']=df['year'].astype(int)
```

```python
df.head()
```

```python
df['Dep_hour']=df['Dep_Time'].str.split(':').str[0]
df['Dep_minute']=df['Dep_Time'].str.split(':').str[1]
df['Dep_hour']=df['Dep_hour'].astype(int)
df['Dep_minute']=df['Dep_minute'].astype(int)
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df['arraival_hour']=df['Arrival_Time'].str.split(':').str[0]
df['arraival_minute']=df['Arrival_Time'].str.split(':').str[1]
df['arraival_hour']=df['arraival_hour'].astype(int)
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
# Extract hours and minutes safely
df['Duration_hours'] = df['Duration'].str.extract(r'(\d+)h')[0].astype(float).fillna(0).astype(int)
df['Duration_minutes'] = df['Duration'].str.extract(r'(\d+)m')[0].astype(float).fillna(0).astype(int)

# Total minutes
df['Total_minutes'] = df['Duration_hours'] * 60 + df['Duration_minutes']

# Drop original column if needed
df.drop('Duration', axis=1, inplace=True)
```
