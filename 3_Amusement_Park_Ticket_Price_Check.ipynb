```python
print('Welcome to the Amusement Park!\n')

height = float(input('Please enter your height in cm: '))
bill = 0

if height > 120:
    print('You are eligible for games and rides!')

    age = int(input('Please enter your age: '))

    if age < 12:
        bill = 5
    elif age <= 18:
        bill = 15
    else:
        bill = 20

    print(f'Your ticket price is ${bill}')

    want_picture = input(
        'Do you need a picture? Type y for Yes and n for No: '
    ).lower()

    if want_picture == 'y':
        bill += 3

    print(f'Your final bill is ${bill}')

else:
    print('You are not eligible for games and rides!')
```
