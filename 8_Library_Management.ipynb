```python
# Calculator Functions

def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

def square(a):
    return a ** 2


# Input Validation Functions

def get_operation():
    while True:
        try:
            operation = int(input(
                "\nChoose Operation:\n"
                "1. Addition\n"
                "2. Subtraction\n"
                "3. Multiplication\n"
                "4. Division\n"
                "5. Square\n"
                "6. View History\n"
                "7. Exit\n"
                "Enter choice: "
            ))

            if operation in range(1, 8):
                return operation

            print("Please choose a number between 1 and 7.")

        except ValueError:
            print("Invalid input. Enter a numeric option.")


def get_number(message):
    while True:
        try:
            return float(input(message))
        except ValueError:
            print("Invalid input. Please enter a valid number.")


# Dictionary Mapping

operations = {
    1: ("Addition", addition),
    2: ("Subtraction", subtraction),
    3: ("Multiplication", multiplication),
    4: ("Division", division),
    5: ("Square", square)
}


# History Storage

history = []

print("Welcome to Simple Calculator")

while True:

    operation = get_operation()

    if operation == 7:
        print("\nThank you for using Simple Calculator!")
        break

    if operation == 6:

        if not history:
            print("\nNo calculations performed yet.")

        else:
            print("\nCalculation History:")
            for item in history:
                print(item)

        continue

    operation_name, operation_function = operations[operation]

    if operation == 5:

        a = get_number("Enter a number: ")

        result = operation_function(a)

        history.append(
            f"{operation_name}: {a}² = {result}"
        )

    else:

        a = get_number("Enter first number: ")
        b = get_number("Enter second number: ")

        result = operation_function(a, b)

        history.append(
            f"{operation_name}: {a} and {b} = {result}"
        )

    print(f"Result: {result}")
```
